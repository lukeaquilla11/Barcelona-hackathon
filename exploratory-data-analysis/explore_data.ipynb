# Loading Datasets

```python
import os, sys

sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
from kloppy import sportscode

import games

path = games.asset_paths()['smart_tagging']
tagging_dataset = sportscode.load(path).to_df()
print('%d instances' % len(tagging_dataset))
# events.head()

tagging_df = pd.DataFrame(tagging_dataset)
tagging_df.sort_values(by='timestamp', inplace=True)
# type(tagging_df['end_timestamp'][0])
tagging_df['end_frame'] = (tagging_df['end_timestamp'].dt.total_seconds()*25).round().astype(int)
tagging_df['start_frame'] = (tagging_df['timestamp'].dt.total_seconds()*25).round().astype(int)
tagging_df.sort_values(by='start_frame')
```

## Tracking ATD

```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
import matplotlib.pyplot as plt
from kloppy import metrica

import games

pd.set_option('display.max_columns', None)

paths = games.asset_paths(source='tactical')

# The full file is 199,319 frames. Enough here to reach the first-half kickoff at 11185.
tracking_dataset = metrica.load_tracking_epts(meta_data=paths['atd_meta'],
                                     raw_data=paths['atd_raw']
                                    )
print('%d frames loaded' % len(tracking_dataset.records))

tracking_df = tracking_dataset.to_df()
tracking_df.head()

tracking_df = pd.DataFrame(tracking_df)
```

```python
import pandas as pd

pd.set_option('display.max_rows', None)

# 1. Create a boolean mask for where ball coordinates are missing
is_missing = tracking_df["ball_x"].isna()

# 2. Group consecutive missing rows together using cumsum()
# This creates a unique ID for each continuous block of missing data
missing_blocks = (is_missing != is_missing.shift()).cumsum()

# 3. Filter for only the missing rows and aggregate their start and end frames
ball_missing_intervals = (
    tracking_df[is_missing]
    .groupby(missing_blocks)
    .agg(
        start_frame=('frame_id', 'first'),
        end_frame=('frame_id', 'last'),
        missing_count=('frame_id', 'count')
    )
    .reset_index(drop=True)
)

# View the intervals where the ball disappeared
print(ball_missing_intervals)
```
